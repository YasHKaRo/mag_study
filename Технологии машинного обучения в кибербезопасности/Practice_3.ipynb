## Описание объекта исследования

**Lazarus Group** — это северокорейская APT-группировка, которая связывается с длительными кампаниями кибершпионажа, саботажа и финансово мотивированных атак. В различных публичных источниках группа также упоминается под именами *HIDDEN COBRA*, *Labyrinth Chollima* и рядом других обозначений.

Для Lazarus Group характерно сочетание нескольких направлений деятельности:

- атаки на государственные и оборонные организации;
- атаки на финансовый сектор и криптовалютную инфраструктуру;
- использование социальной инженерии;
- применение вредоносных документов и цепочек начального заражения;
- развитие атаки внутри сети через lateral movement;
- активное использование defense evasion и custom malware.

Почему для практической работы выбрана именно Lazarus Group:
- по этой группировке имеется достаточное количество открытых аналитических материалов;
- у группы выраженный набор TTP, который удобно переводить в поведенческие признаки;
- по Lazarus есть как кампании кибершпионажа, так и финансовые атаки, поэтому можно анализировать широкий профиль поведения.


## Исследование Lazarus Group: цели, инструменты, векторы проникновения и TTP

### Цели атак

Для Lazarus Group характерны следующие цели:

- получение чувствительной информации;
- компрометация оборонных, инженерных и государственных организаций;
- хищение денежных средств;
- атаки на криптовалютные компании и блокчейн-инфраструктуру;
- проведение разрушительных действий и нарушение работы инфраструктуры.

### Типовые инструменты и методы

По открытым описаниям кампаний Lazarus Group часто использует:

- spearphishing-письма;
- вредоносные документы с макросами или шаблонами;
- PowerShell и системные интерпретаторы;
- DLL side-loading;
- scheduled tasks;
- использование `rundll32`, `regsvr32`, `mshta`;
- SMB / RDP / SSH для перемещения по сети;
- custom malware и зашифрованные либо упакованные полезные нагрузки;
- каналы C2 и web/cloud exfiltration.

### Векторы проникновения

Наиболее характерные векторы начального доступа:

- целевые фишинговые письма с вложением;
- фишинговые ссылки;
- контакт через социальные платформы и ложные предложения о работе;
- троянизированные приложения;
- злоупотребление доверием к легитимным инструментам.

### Lateral movement

После первичного закрепления злоумышленник может использовать:

- SMB / Windows Admin Shares;
- RDP;
- SSH;
- компрометацию учетных данных и valid accounts;
- сетевую разведку и использование внутренних прокси.

### Обход защитных механизмов

Типовые методы defense evasion:

- маскировка вредоносной активности под легитимные процессы;
- упаковка и шифрование полезной нагрузки;
- использование signed binaries и LOLBin-подобных сценариев;
- scheduled tasks и сервисы для persistence;
- удаление артефактов и снижение заметности следов.

## Жизненный цикл атаки (Kill Chain) для Lazarus Group

### Reconnaissance
Злоумышленник собирает сведения о сотрудниках, подразделениях, технологиях и внешней инфраструктуре цели.  
Для Lazarus Group характерно изучение конкретных сотрудников и ролей, особенно в оборонной и финансовой сфере.

### Weaponization
Подготавливаются вредоносные документы, архивы, скрипты, троянизированные приложения или цепочки с side-loading.

### Delivery
Доставка осуществляется через:
- phishing attachment;
- phishing link;
- социальные платформы;
- ложные вакансии и предложения сотрудничества;
- вредоносные приложения.

### Exploitation
Жертва открывает документ, запускает файл или переходит по ссылке. Активируются макросы, скрипты, шаблоны, PowerShell-команды или DLL.

### Installation
Злоумышленник закрепляется на системе:
- scheduled tasks;
- новые сервисы;
- side-loading;
- запуск через системные бинарные файлы.

### Command and Control
Организуется обмен с C2-инфраструктурой, иногда с резервными каналами связи.

### Actions on Objectives
В зависимости от кампании это могут быть:
- кража данных;
- перемещение по сети;
- кража учетных данных;
- хищение криптоактивов;
- разрушительные действия;
- подготовка следующего этапа атаки.

## Описание датасета

В данной работе используется **сгенерированный датасет на 5000 объектов**.
Проблема в том, что реальные открытые датасеты:
- редко содержат явную привязку к конкретной APT-группе;

- не дают корректной разметки именно под Lazarus Group.

Поэтому строится поведенческий датасет, где:
- положительный класс (`target = 1`) отражает события, похожие на профиль Lazarus Group;
- отрицательный класс (`target = 0`) отражает обычные события и нерелевантную вредоносную активность.

### Структура датасета
Один объект датасета — это условная агрегированная активность хоста / сессии / цепочки событий, приведенная к единому набору признаков.

### Ограничения
Такой датасет подходит для исследовательской и учебной постановки, но не заменяет промышленную модель, обученную на реальной телеметрии организации.


```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier

RANDOM_STATE = 123
np.random.seed(RANDOM_STATE)
```

```python
feature_columns = [
    "phish_attachment",
    "phish_link",
    "social_contact",
    "macro_exec",
    "powershell_exec",
    "signed_binary_proxy",
    "dll_sideload",
    "scheduled_task",
    "new_service",
    "credential_access",
    "valid_accounts_use",
    "smb_lateral",
    "rdp_lateral",
    "ssh_lateral",
    "network_discovery",
    "defense_evasion",
    "packed_payload",
    "c2_beaconing",
    "cloud_exfil",
    "crypto_targeting",
    "failed_login_burst",
    "outbound_conn_count",
    "rare_parent_child_score",
    "offhours_activity_score",
    "archive_attachment_kb",
    "script_chain_depth"
]

binary_features = [
    "phish_attachment",
    "phish_link",
    "social_contact",
    "macro_exec",
    "powershell_exec",
    "signed_binary_proxy",
    "dll_sideload",
    "scheduled_task",
    "new_service",
    "credential_access",
    "valid_accounts_use",
    "smb_lateral",
    "rdp_lateral",
    "ssh_lateral",
    "network_discovery",
    "defense_evasion",
    "packed_payload",
    "c2_beaconing",
    "cloud_exfil",
    "crypto_targeting"
]
```

## Генерация датасета

Ниже формируется датасет, в котором положительный класс специально моделируется так, чтобы отражать поведенческий профиль Lazarus Group.

Логика генерации:
- для класса `1` повышаются вероятности фишинга, макросов, PowerShell, scheduled tasks, lateral movement, C2, defense evasion и cloud exfiltration;
- для класса `0` эти признаки встречаются заметно реже и менее согласованно;
- добавляются шум и вариативность, чтобы задача не была тривиальной.


```python
def generate_binary(prob, size):
    return np.random.binomial(1, prob, size)

def generate_lazarus_dataset(n_samples=5000, positive_share=0.32):
    n_pos = int(n_samples * positive_share)
    n_neg = n_samples - n_pos

    # Положительный класс: события, похожие на Lazarus
    pos = pd.DataFrame({
        "phish_attachment": generate_binary(0.72, n_pos),
        "phish_link": generate_binary(0.48, n_pos),
        "social_contact": generate_binary(0.44, n_pos),
        "macro_exec": generate_binary(0.63, n_pos),
        "powershell_exec": generate_binary(0.67, n_pos),
        "signed_binary_proxy": generate_binary(0.58, n_pos),
        "dll_sideload": generate_binary(0.39, n_pos),
        "scheduled_task": generate_binary(0.61, n_pos),
        "new_service": generate_binary(0.33, n_pos),
        "credential_access": generate_binary(0.54, n_pos),
        "valid_accounts_use": generate_binary(0.49, n_pos),
        "smb_lateral": generate_binary(0.46, n_pos),
        "rdp_lateral": generate_binary(0.31, n_pos),
        "ssh_lateral": generate_binary(0.22, n_pos),
        "network_discovery": generate_binary(0.64, n_pos),
        "defense_evasion": generate_binary(0.69, n_pos),
        "packed_payload": generate_binary(0.57, n_pos),
        "c2_beaconing": generate_binary(0.74, n_pos),
        "cloud_exfil": generate_binary(0.41, n_pos),
        "crypto_targeting": generate_binary(0.28, n_pos),
        "failed_login_burst": np.random.poisson(6, n_pos),
        "outbound_conn_count": np.random.poisson(18, n_pos) + 4,
        "rare_parent_child_score": np.round(np.random.beta(5, 2, n_pos), 3),
        "offhours_activity_score": np.round(np.random.beta(4, 2, n_pos), 3),
        "archive_attachment_kb": np.random.randint(120, 2400, n_pos),
        "script_chain_depth": np.random.randint(2, 8, n_pos),
        "target": 1
    })

    # Отрицательный класс: обычные или нерелевантные события
    neg = pd.DataFrame({
        "phish_attachment": generate_binary(0.36, n_neg),
        "phish_link": generate_binary(0.22, n_neg),
        "social_contact": generate_binary(0.14, n_neg),
        "macro_exec": generate_binary(0.18, n_neg),
        "powershell_exec": generate_binary(0.36, n_neg),
        "signed_binary_proxy": generate_binary(0.20, n_neg),
        "dll_sideload": generate_binary(0.15, n_neg),
        "scheduled_task": generate_binary(0.14, n_neg),
        "new_service": generate_binary(0.12, n_neg),
        "credential_access": generate_binary(0.20, n_neg),
        "valid_accounts_use": generate_binary(0.34, n_neg),
        "smb_lateral": generate_binary(0.18, n_neg),
        "rdp_lateral": generate_binary(0.16, n_neg),
        "ssh_lateral": generate_binary(0.12, n_neg),
        "network_discovery": generate_binary(0.36, n_neg),
        "defense_evasion": generate_binary(0.24, n_neg),
        "packed_payload": generate_binary(0.18, n_neg),
        "c2_beaconing": generate_binary(0.34, n_neg),
        "cloud_exfil": generate_binary(0.20, n_neg),
        "crypto_targeting": generate_binary(0.10, n_neg),
        "failed_login_burst": np.random.poisson(3, n_neg),
        "outbound_conn_count": np.random.poisson(12, n_neg) + 1,
        "rare_parent_child_score": np.round(np.random.beta(2, 5, n_neg), 3),
        "offhours_activity_score": np.round(np.random.beta(2, 5, n_neg), 3),
        "archive_attachment_kb": np.random.randint(80, 1000, n_neg),
        "script_chain_depth": np.random.randint(1, 4, n_neg),
        "target": 0
    })

    df = pd.concat([pos, neg], ignore_index=True)

    # Небольшой шум: часть признаков случайно инвертируется у малой доли строк
    noise_idx = np.random.choice(df.index, size=int(0.04 * len(df)), replace=False)
    for col in binary_features:
        flip_mask = np.random.binomial(1, 0.08, len(noise_idx)).astype(bool)
        idx_to_flip = pd.Index(noise_idx[flip_mask])
        df.loc[idx_to_flip, col] = 1 - df.loc[idx_to_flip, col]

    df = df.sample(frac=1, random_state=RANDOM_STATE).reset_index(drop=True)
    return df

df = generate_lazarus_dataset(n_samples=5000, positive_share=0.32)
df.head()
```

### Краткая характеристика датасета

В рамках практической работы был сформирован сгенерированный датасет объемом 5000 объектов.  
Каждый объект отражает агрегированную активность, описанную набором бинарных и числовых признаков.  
Положительный класс соответствует активности, поведенчески близкой к Lazarus Group, отрицательный — обычным либо нерелевантным событиям.  

При генерации положительного класса были повышены вероятности признаков, характерных для Lazarus Group: целевой фишинг, выполнение макросов, запуск PowerShell, scheduled tasks, defense evasion, C2-активность, lateral movement и cloud exfiltration.  
Дополнительно использовались числовые признаки, характеризующие аномальную активность: число внешних соединений, активность в нерабочее время, редкие parent-child связи процессов, глубина цепочки скриптов и размер архивных вложений.  

Такой подход позволяет получить информативную обучающую выборку, которая отражает не отдельный индикатор, а согласованный профиль поведения атакующей группировки.


```python
summary_table = pd.DataFrame({
    "feature": feature_columns,
    "mean_target_0": [df[df["target"] == 0][c].mean() for c in feature_columns],
    "mean_target_1": [df[df["target"] == 1][c].mean() for c in feature_columns],
})
summary_table["diff"] = summary_table["mean_target_1"] - summary_table["mean_target_0"]
summary_table.sort_values("diff", ascending=False).head(15)
```

```python
X = df[feature_columns]
y = df["target"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    random_state=RANDOM_STATE,
    stratify=y
)

models = {
    "RandomForest": RandomForestClassifier(
        n_estimators=250,
        max_depth=10,
        min_samples_leaf=3,
        random_state=RANDOM_STATE,
        n_jobs=-1
    ),
    "GradientBoosting": GradientBoostingClassifier(
        n_estimators=180,
        learning_rate=0.08,
        max_depth=3,
        random_state=RANDOM_STATE
    )
}
```

## Обучение и сравнение моделей

Согласно заданию, желательно рассмотреть несколько моделей и сравнить результаты.  
В работе используются:

- **Logistic Regression** — как базовая интерпретируемая модель;
- **Random Forest** — как устойчивая модель для табличных признаков;
- **Gradient Boosting** — как более гибкая ансамблевая модель.

Сравнение проводится по метрикам:
- Accuracy;
- Precision;
- Recall;
- F1-score;
- ROC-AUC.


```python
results = []

trained_models = {}

for name, model in models.items():
    model.fit(X_train, y_train)
    trained_models[name] = model

    y_pred = model.predict(X_test)
    if hasattr(model, "predict_proba"):
        y_prob = model.predict_proba(X_test)[:, 1]
    else:
        y_prob = y_pred

    results.append({
        "model": name,
        "accuracy": round(accuracy_score(y_test, y_pred), 4),
        "precision": round(precision_score(y_test, y_pred), 4),
        "recall": round(recall_score(y_test, y_pred), 4),
        "f1": round(f1_score(y_test, y_pred), 4),
        "roc_auc": round(roc_auc_score(y_test, y_prob), 4)
    })

results_df = pd.DataFrame(results).sort_values("roc_auc", ascending=False).reset_index(drop=True)
results_df
```

### Обоснование выбора модели

Для решения задачи были рассмотрены две модели: Random Forest и Gradient Boosting.  

Random Forest хорошо подходит для табличных данных, устойчив к шуму и позволяет учитывать сложные комбинации поведенческих признаков.  
Gradient Boosting также хорошо работает на задачах классификации и способен выявлять нелинейные зависимости, однако может быть более чувствителен к настройкам.  

В данной задаче приоритет отдается модели, которая лучше выявляет согласованный профиль TTP, а не одиночные индикаторы. Поэтому на практике наибольший интерес представляют ансамблевые модели.


```python
best_model_name = results_df.iloc[0]["model"]
best_model = trained_models[best_model_name]

best_pred = best_model.predict(X_test)
best_prob = best_model.predict_proba(X_test)[:, 1]

print("Best model:", best_model_name)
print()
print(classification_report(y_test, best_pred, digits=4))
print("Confusion matrix:")
print(confusion_matrix(y_test, best_pred))
```

```python
if best_model_name == "RandomForest":
    importance_df = pd.DataFrame({
        "feature": feature_columns,
        "importance": best_model.feature_importances_
    }).sort_values("importance", ascending=False).reset_index(drop=True)
else:
    if hasattr(best_model, "feature_importances_"):
        importance_df = pd.DataFrame({
            "feature": feature_columns,
            "importance": best_model.feature_importances_
        }).sort_values("importance", ascending=False).reset_index(drop=True)
    else:
        importance_df = pd.DataFrame({
            "feature": feature_columns,
            "importance": np.abs(best_model.coef_[0])
        }).sort_values("importance", ascending=False).reset_index(drop=True)

importance_df.head(15)
```

## Анализ ошибок модели

Анализ ошибок необходим, чтобы понять, где модель ошибается:

- **False Positive**: модель считает событие похожим на Lazarus, хотя оно относится к отрицательному классу;
- **False Negative**: модель пропускает активность, похожую на Lazarus.

Для SOC-практики особенно критичны false negative, так как это фактически пропущенная атака.  
Однако большое число false positive тоже нежелательно, потому что перегружает аналитиков.

Ниже проводится сравнение FP и FN.


```python
error_df = X_test.copy()
error_df["y_true"] = y_test.values
error_df["y_pred"] = best_pred
error_df["y_prob"] = best_prob

fp_df = error_df[(error_df["y_true"] == 0) & (error_df["y_pred"] == 1)].copy()
fn_df = error_df[(error_df["y_true"] == 1) & (error_df["y_pred"] == 0)].copy()

len(fp_df), len(fn_df)
```

```python
fp_profile = fp_df[feature_columns].mean().sort_values(ascending=False).head(10)
fn_profile = fn_df[feature_columns].mean().sort_values(ascending=False).head(10)

fp_profile, fn_profile
```

По результатам эксперимента были проанализированы ошибки первого и второго рода.  
Ложноположительные срабатывания возникают в случаях, когда в объекте присутствует часть признаков, характерных для Lazarus Group, но отсутствует полноценная согласованная цепочка атаки. Например, это может быть PowerShell-активность или scheduled task без выраженного lateral movement и без устойчивого C2-поведения.  

Ложноотрицательные ошибки возникают тогда, когда активность злоумышленника проявляется частично или замаскирована под обычную административную активность.  
Наиболее сложными для модели являются пограничные случаи, в которых отдельные признаки attack-chain распределены неравномерно: например, есть признаки initial access и defense evasion, но слабо выражены exfiltration и network discovery.  

Таким образом, качество модели зависит не только от наличия отдельных индикаторов, но и от того, насколько полно в объекте отражен связанный профиль TTP.


## Интерпретация результатов

В рамках учебной задачи модель демонстрирует способность выявлять подобные паттерны на ранних этапах, особенно если уже зафиксированы признаки phishing, script execution и defense evasion.
