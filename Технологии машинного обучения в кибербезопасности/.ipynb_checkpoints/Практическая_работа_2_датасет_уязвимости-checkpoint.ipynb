# Практическая работа №2.1  
## Формирование датасета для предсказания уязвимостей по исходному коду и истории разработки

Этот ноутбук разбит на короткие логические блоки, чтобы его было удобно запускать по ячейкам в Jupyter.

Что делает ноутбук:
1. Загружает две исходные таблицы:
   - `output_files.csv` — файлы, связанные с исправлениями уязвимостей;
   - `output_files_norm.csv` — файлы без такой разметки.
2. Формирует единую таблицу с меткой класса.
3. Проводит предварительную очистку и анализ.
4. Строит базовые признаки по метаданным файла.
5. При необходимости догружает исходный код по `raw_url` и считает метрики кода.
6. При необходимости догружает историю разработки файла через GitHub API.
7. Сохраняет итоговый датасет для последующего ML.

> Базовая идея полностью соответствует постановке практической работы: разметка по файлам/компонентам, выделение признаков, проверка данных, формирование итогового датасета.

## 5. Предобработка: парсинг `patch_signif`

В исходных таблицах `patch_signif` хранится как строковое представление списка diff-ханков.  
Преобразуем его в количество значимых изменений.

````python

def safe_parse_patch_signif(value):
    if pd.isna(value):
        return []
    if isinstance(value, list):
        return value
    try:
        parsed = ast.literal_eval(value)
        return parsed if isinstance(parsed, list) else []
    except Exception:
        return []

df["patch_signif_list"] = df["patch_signif"].apply(safe_parse_patch_signif)
df["patch_count"] = df["patch_signif_list"].apply(len)

display(df[["patch_signif", "patch_signif_list", "patch_count"]].head())
````

## 9. Формирование признаков по метаданным файла

Это базовый и полностью воспроизводимый слой признаков, который можно получить даже без скачивания исходного кода:
- длина пути;
- глубина вложенности;
- признаки по имени файла и каталогу;
- размер исправления через `patch_count`;
- простые семантические индикаторы по пути.

````python

def build_metadata_features(frame: pd.DataFrame) -> pd.DataFrame:
    out = frame.copy()

    out["file_basename"] = out["file_name"].str.split("/").str[-1]
    out["dir_path"] = out["file_name"].str.rsplit("/", n=1).str[0]
    out["dir_path"] = out["dir_path"].where(out["dir_path"] != out["file_basename"], "")

    out["path_depth"] = out["file_name"].str.count("/")
    out["path_len"] = out["file_name"].str.len()
    out["filename_len"] = out["file_basename"].str.len()
    out["token_count_in_path"] = out["file_name"].str.count(r"[/_.-]") + 1
    out["raw_url_len"] = out["raw_url"].astype(str).str.len()

    lower_path = out["file_name"].str.lower()

    out["is_test_like"] = lower_path.str.contains(r"(^|/)(test|tests|testing)(/|$)|_test\.|test_", regex=True).astype(int)
    out["is_internal_path"] = lower_path.str.contains(r"(^|/)internal(/|$)", regex=True).astype(int)
    out["is_pkg_path"] = lower_path.str.contains(r"(^|/)pkg(/|$)", regex=True).astype(int)
    out["is_cmd_path"] = lower_path.str.contains(r"(^|/)cmd(/|$)", regex=True).astype(int)

    keyword_patterns = {
        "has_auth_kw": r"auth|login|token|session|permission|acl|rbac",
        "has_crypto_kw": r"crypto|cipher|tls|ssl|x509|jwt|oauth|saml|hash|sign",
        "has_network_kw": r"http|grpc|tcp|udp|socket|proxy|request|server|client",
        "has_storage_kw": r"db|sql|redis|mongo|storage|cache|repo|repository",
        "has_parser_kw": r"parse|parser|decode|encode|marshal|unmarshal|yaml|json|xml",
        "has_input_kw": r"input|user|request|form|query|param|args",
        "has_exec_kw": r"exec|command|shell|os|syscall|runtime",
        "has_security_kw": r"security|vuln|sanitize|escape|csrf|xss|injection|secret",
    }

    for col, pattern in keyword_patterns.items():
        out[col] = lower_path.str.contains(pattern, regex=True).astype(int)

    return out

df_feat = build_metadata_features(df)
display(df_feat.head())
````

## 10. Агрегированные признаки по репозиторию и коммиту исправления

Это приближенный аналог исторических/контекстных признаков:
- сколько файлов в датасете приходится на репозиторий;
- сколько файлов менялось в рамках одного fix-коммита;
- сколько объектов в той же директории;
- как часто этот файл встречается в разметке.

````python

repo_counts = df_feat.groupby("repo_key").size().rename("repo_rows_in_dataset")
owner_counts = df_feat.groupby("owner").size().rename("owner_rows_in_dataset")
fix_counts = df_feat.groupby(["repo_key", "vulnerable_hash"]).size().rename("files_in_same_fix_commit")
dir_counts = df_feat.groupby(["repo_key", "dir_path"]).size().rename("rows_in_same_dir_in_dataset")
file_counts = df_feat.groupby(["repo_key", "file_name"]).size().rename("same_file_occurrences_in_dataset")
repo_pos_ratio = df_feat.groupby("repo_key")["label"].mean().rename("repo_positive_ratio_in_dataset")

df_feat = df_feat.join(repo_counts, on="repo_key")
df_feat = df_feat.join(owner_counts, on="owner")
df_feat = df_feat.join(fix_counts, on=["repo_key", "vulnerable_hash"])
df_feat = df_feat.join(dir_counts, on=["repo_key", "dir_path"])
df_feat = df_feat.join(file_counts, on=["repo_key", "file_name"])
df_feat = df_feat.join(repo_pos_ratio, on="repo_key")

display(df_feat.head())
````

## 13. Опционально: скачивание исходного кода по `raw_url`

Для большого датасета лучше качать не все сразу, а батчами.  
Ниже функция с локальным кэшем.

Если GitHub ограничивает запросы, можно использовать токен:
```python
GITHUB_TOKEN = "ghp_..."
```

````python

# Раскомментируйте, если нужен токен GitHub:
# GITHUB_TOKEN = "YOUR_GITHUB_TOKEN"

try:
    import requests
except ImportError:
    requests = None
    print("requests не установлен. Установите: pip install requests")

def make_cache_name(raw_url: str) -> Path:
    file_hash = hashlib.md5(raw_url.encode("utf-8")).hexdigest()
    return RAW_CACHE_DIR / f"{file_hash}.txt"

def download_raw_text(raw_url: str, use_cache: bool = True, sleep_sec: float = 0.0) -> str | None:
    if requests is None:
        return None

    cache_file = make_cache_name(raw_url)
    if use_cache and cache_file.exists():
        return cache_file.read_text(encoding="utf-8", errors="ignore")

    headers = {}
    if "GITHUB_TOKEN" in globals() and GITHUB_TOKEN:
        headers["Authorization"] = f"Bearer {GITHUB_TOKEN}"

    try:
        resp = requests.get(raw_url, headers=headers, timeout=30)
        if resp.status_code == 200:
            text = resp.text
            cache_file.write_text(text, encoding="utf-8", errors="ignore")
            if sleep_sec:
                time.sleep(sleep_sec)
            return text
        else:
            print(f"Не удалось скачать {raw_url}: {resp.status_code}")
            return None
    except Exception as e:
        print(f"Ошибка при скачивании {raw_url}: {e}")
        return None
````

## 14. Опционально: выбор подвыборки для скачивания

## 16. Опционально: функции для расчета метрик исходного кода

Так как в ваших таблицах сейчас все файлы имеют расширение `.go`, ниже сделан упор на Go.  
Часть метрик универсальная, часть — Go-специфичная.

````python

COMMENT_RE_LINE = re.compile(r"^\s*//")
FUNC_RE_GO = re.compile(r"^\s*func\s+")
STRUCT_RE_GO = re.compile(r"\btype\s+\w+\s+struct\b")
INTERFACE_RE_GO = re.compile(r"\btype\s+\w+\s+interface\b")

def compute_brace_nesting(lines):
    depth = 0
    max_depth = 0
    for line in lines:
        opens = line.count("{")
        closes = line.count("}")
        depth += opens
        max_depth = max(max_depth, depth)
        depth -= closes
        depth = max(depth, 0)
    return max_depth

def extract_code_metrics(text: str) -> dict:
    if not isinstance(text, str) or not text:
        return {
            "code_bytes": np.nan,
            "lines_total": np.nan,
            "lines_nonempty": np.nan,
            "lines_comment": np.nan,
            "comment_ratio": np.nan,
            "avg_line_len": np.nan,
            "max_line_len": np.nan,
            "func_count": np.nan,
            "struct_count": np.nan,
            "interface_count": np.nan,
            "if_count": np.nan,
            "for_count": np.nan,
            "switch_count": np.nan,
            "select_count": np.nan,
            "return_count": np.nan,
            "defer_count": np.nan,
            "go_stmt_count": np.nan,
            "panic_count": np.nan,
            "import_count": np.nan,
            "string_literal_count": np.nan,
            "max_brace_nesting": np.nan,
        }

    lines = text.splitlines()
    stripped = [line.strip() for line in lines]

    lines_total = len(lines)
    lines_nonempty = sum(bool(x) for x in stripped)
    lines_comment = sum(bool(COMMENT_RE_LINE.match(x)) for x in lines)

    avg_line_len = np.mean([len(x) for x in lines]) if lines else 0
    max_line_len = max((len(x) for x in lines), default=0)

    metrics = {
        "code_bytes": len(text.encode("utf-8", errors="ignore")),
        "lines_total": lines_total,
        "lines_nonempty": lines_nonempty,
        "lines_comment": lines_comment,
        "comment_ratio": lines_comment / lines_total if lines_total else 0.0,
        "avg_line_len": avg_line_len,
        "max_line_len": max_line_len,
        "func_count": sum(bool(FUNC_RE_GO.match(x)) for x in lines),
        "struct_count": len(STRUCT_RE_GO.findall(text)),
        "interface_count": len(INTERFACE_RE_GO.findall(text)),
        "if_count": len(re.findall(r"\bif\b", text)),
        "for_count": len(re.findall(r"\bfor\b", text)),
        "switch_count": len(re.findall(r"\bswitch\b", text)),
        "select_count": len(re.findall(r"\bselect\b", text)),
        "return_count": len(re.findall(r"\breturn\b", text)),
        "defer_count": len(re.findall(r"\bdefer\b", text)),
        "go_stmt_count": len(re.findall(r"\bgo\b", text)),
        "panic_count": len(re.findall(r"\bpanic\b", text)),
        "import_count": len(re.findall(r"\bimport\b", text)),
        "string_literal_count": len(re.findall(r'"[^"\\]*(?:\\.[^"\\]*)*"', text)),
        "max_brace_nesting": compute_brace_nesting(lines),
    }
    return metrics
````

# Дополнительный блок: признаки истории разработки

Для практической работы полезно показать, что вы учитываете и историю разработки файла.

Ниже — заготовка, которую можно использовать с GitHub API:
- число последних коммитов файла;
- число авторов;
- время между изменениями;
- доля security-like commit messages.

## 19. Опционально: функция для сбора истории файла из GitHub API

Важно: для большого объема данных этот шаг лучше делать батчами и с токеном GitHub.

````python

def github_commits_url(owner: str, repo: str, file_path: str) -> str:
    from urllib.parse import quote
    quoted_path = quote(file_path)
    return f"https://api.github.com/repos/{owner}/{repo}/commits?path={quoted_path}&per_page=100"

def fetch_file_commit_history(owner: str, repo: str, file_path: str):
    if requests is None:
        return None

    url = github_commits_url(owner, repo, file_path)
    headers = {"Accept": "application/vnd.github+json"}
    if "GITHUB_TOKEN" in globals() and GITHUB_TOKEN:
        headers["Authorization"] = f"Bearer {GITHUB_TOKEN}"

    try:
        resp = requests.get(url, headers=headers, timeout=30)
        if resp.status_code == 200:
            return resp.json()
        else:
            print(f"Ошибка {resp.status_code}: {owner}/{repo} {file_path}")
            return None
    except Exception as e:
        print(f"Ошибка запроса истории: {e}")
        return None

def summarize_commit_history(commits_json):
    if not commits_json:
        return {
            "hist_commit_count": np.nan,
            "hist_author_count": np.nan,
            "hist_days_span": np.nan,
            "hist_avg_days_between_commits": np.nan,
            "hist_security_msg_count": np.nan,
        }

    commit_dates = []
    authors = set()
    security_msg_count = 0

    for item in commits_json:
        commit = item.get("commit", {})
        author = commit.get("author", {}) or {}
        date_str = author.get("date")
        if date_str:
            commit_dates.append(pd.to_datetime(date_str, utc=True, errors="coerce"))
        email = author.get("email")
        name = author.get("name")
        authors.add((name, email))

        message = (commit.get("message") or "").lower()
        if re.search(r"fix|security|vuln|cve|overflow|xss|csrf|inject|panic|dos|auth", message):
            security_msg_count += 1

    commit_dates = [x for x in commit_dates if pd.notna(x)]
    commit_dates = sorted(commit_dates)

    if len(commit_dates) >= 2:
        deltas = np.diff(commit_dates).astype("timedelta64[s]").astype(float) / 86400.0
        avg_delta = float(np.mean(deltas))
        span_days = float((commit_dates[-1] - commit_dates[0]).total_seconds() / 86400.0)
    else:
        avg_delta = np.nan
        span_days = 0.0 if len(commit_dates) == 1 else np.nan

    return {
        "hist_commit_count": len(commits_json),
        "hist_author_count": len(authors),
        "hist_days_span": span_days,
        "hist_avg_days_between_commits": avg_delta,
        "hist_security_msg_count": security_msg_count,
    }
````

## 20. Опционально: пример сбора истории для небольшой подвыборки

````python

HISTORY_SAMPLE_SIZE = 50

history_sample = df_feat.sample(
    n=min(HISTORY_SAMPLE_SIZE, len(df_feat)),
    random_state=42
)[["sample_id", "owner", "repo", "file_name"]].copy()

display(history_sample.head())
````

````python

if requests is not None:
    history_rows = []

    for _, row in history_sample.iterrows():
        commits_json = fetch_file_commit_history(row["owner"], row["repo"], row["file_name"])
        metrics = summarize_commit_history(commits_json)
        history_rows.append({
            "sample_id": row["sample_id"],
            **metrics
        })
        time.sleep(0.2)

    history_metrics_df = pd.DataFrame(history_rows)
    display(history_metrics_df.head())
else:
    print("Пропущено: requests не установлен")
````

## 23. Разбиение train/test без утечки по репозиториям

Лучше делить не случайно по строкам, а группами по репозиториям.  
Иначе одна и та же кодовая база попадет и в train, и в test, что даст слишком оптимистичную оценку.

````python

from sklearn.model_selection import GroupShuffleSplit

groups = baseline_dataset["repo_key"]

gss = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
train_idx, test_idx = next(gss.split(X, y, groups=groups))

X_train = X.iloc[train_idx].reset_index(drop=True)
X_test = X.iloc[test_idx].reset_index(drop=True)
y_train = y.iloc[train_idx].reset_index(drop=True)
y_test = y.iloc[test_idx].reset_index(drop=True)

print("Train:", X_train.shape, y_train.shape)
print("Test :", X_test.shape, y_test.shape)

print("Train positive ratio:", round(y_train.mean(), 4))
print("Test positive ratio :", round(y_test.mean(), 4))
````
