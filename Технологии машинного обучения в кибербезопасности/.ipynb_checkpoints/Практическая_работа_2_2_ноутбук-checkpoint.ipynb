# Практическая работа №2.2  
## Предсказание уязвимостей на основе данных об исходном коде и истории ПО

В этом ноутбуке выполняются пункты задания:

1. обучение модели на ранее сформированном датасете;
2. выбор нового проекта, не использовавшегося при обучении;
3. применение модели к новому проекту;
4. получение вероятностей уязвимости для файлов;
5. проверка, попали ли реальные уязвимые файлы в наиболее рискованные;
6. оценка ранжирования;
7. формирование итоговой таблицы результатов;
8. вывод о способности модели переноситься на новый проект.

**Важно.** В имеющихся данных уже есть множество GitHub-репозиториев. Поэтому новый проект здесь выбирается как **полностью отложенный репозиторий**: он не участвует в обучении, а используется только для тестирования. Это корректный сценарий проверки переносимости модели на unseen project.

## 1. Импорты

````
import ast
import re
from pathlib import Path

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    average_precision_score,
    classification_report,
    precision_score,
    recall_score,
    roc_auc_score,
)
````

## 2. Загрузка данных

Ожидаются два CSV-файла:

- `output_files.csv` — файлы, связанные с исправлениями уязвимостей;
- `output_files_norm.csv` — обычные файлы без известной уязвимости.

Для обучения вводится бинарная метка:
- `label = 1` — уязвимый файл;
- `label = 0` — обычный файл.

````
VULN_PATH = "output_files.csv"
NORM_PATH = "output_files_norm.csv"

df_vuln = pd.read_csv(VULN_PATH)
df_norm = pd.read_csv(NORM_PATH)

df_vuln["label"] = 1
df_norm["label"] = 0

print("Размер vulnerable:", df_vuln.shape)
print("Размер normal:", df_norm.shape)
display(df_vuln.head(3))
display(df_norm.head(3))
````

## 3. Объединение и минимальная очистка

````
df = pd.concat([df_vuln, df_norm], ignore_index=True)

# Удаляем точные дубли по ключевым полям
dedup_cols = ["owner", "repo", "vulnerable_hash", "file_name", "raw_url", "label"]
dedup_cols = [c for c in dedup_cols if c in df.columns]

before = len(df)
df = df.drop_duplicates(subset=dedup_cols).copy()
after = len(df)

# Переименуем колонки к более удобным рабочим именам
df = df.rename(columns={
    "file_name": "path",
    "vulnerable_hash": "fix_commit_hash",
})

print(f"Строк до очистки: {before}")
print(f"Строк после очистки: {after}")
print(f"Удалено дублей: {before - after}")

display(df.head(5))
````

## 4. Быстрый обзор датасета

````
print("Колонки:")
print(df.columns.tolist())
print()

print("Распределение классов:")
display(df["label"].value_counts(dropna=False).rename_axis("label").reset_index(name="count"))

print("Число репозиториев:", df[["owner", "repo"]].drop_duplicates().shape[0])
print("Число владельцев:", df["owner"].nunique())
````

````
class_counts = df["label"].value_counts().sort_index()

plt.figure(figsize=(5, 3))
plt.bar(class_counts.index.astype(str), class_counts.values)
plt.title("Распределение классов")
plt.xlabel("label")
plt.ylabel("Количество файлов")
plt.show()
````

## 5. Формирование признаков

Ниже строятся 24 базовых признака. Они опираются только на те поля, которые уже есть в таблицах:

- путь к файлу;
- raw URL;
- hash исправляющего коммита;
- принадлежность к репозиторию;
- информация о patch-фрагментах.

Это удобно, потому что ноутбук не требует дополнительных обращений к GitHub API.

````
def parse_patch_count(x):
    if pd.isna(x):
        return 0
    try:
        value = ast.literal_eval(x) if isinstance(x, str) else x
        return len(value) if isinstance(value, (list, tuple)) else 0
    except Exception:
        return 0


def tokenize_path(path: str):
    return [tok for tok in re.split(r"[\\/._\-]+", str(path).lower()) if tok]


def has_kw(path: str, keywords):
    p = str(path).lower()
    return int(any(k in p for k in keywords))
````

````
df["dir"] = df["path"].apply(lambda x: str(Path(str(x)).parent))

df["patch_count"] = df["patch_signif"].apply(parse_patch_count)
df["path_depth"] = df["path"].apply(lambda x: len(Path(str(x)).parts))
df["path_len"] = df["path"].astype(str).str.len()
df["filename_len"] = df["path"].apply(lambda x: len(Path(str(x)).name))
df["token_count_in_path"] = df["path"].apply(lambda x: len(tokenize_path(x)))
df["raw_url_len"] = df["raw_url"].astype(str).str.len()

# Признаки по типу пути
df["is_test_like"] = (
    df["path"]
    .astype(str)
    .str.contains(r"(?:^|/)(?:test|tests|testing|spec|specs|mock|mocks)(?:/|$)|_test\.go$", case=False, regex=True, na=False)
    .astype(int)
)
df["is_internal_path"] = df["path"].astype(str).str.contains(r"(?:^|/)internal(?:/|$)", regex=True, na=False).astype(int)
df["is_pkg_path"] = df["path"].astype(str).str.contains(r"(?:^|/)pkg(?:/|$)", regex=True, na=False).astype(int)
df["is_cmd_path"] = df["path"].astype(str).str.contains(r"(?:^|/)cmd(?:/|$)", regex=True, na=False).astype(int)

# Семантические ключевые слова в пути
df["has_auth_kw"] = df["path"].apply(lambda x: has_kw(x, ["auth", "login", "token", "session", "oauth"]))
df["has_crypto_kw"] = df["path"].apply(lambda x: has_kw(x, ["crypto", "tls", "ssl", "x509", "cipher", "cert"]))
df["has_network_kw"] = df["path"].apply(lambda x: has_kw(x, ["http", "grpc", "net", "socket", "tcp", "udp", "proxy"]))
df["has_storage_kw"] = df["path"].apply(lambda x: has_kw(x, ["db", "store", "storage", "cache", "redis", "sql", "mysql", "postgres"]))
df["has_parser_kw"] = df["path"].apply(lambda x: has_kw(x, ["parse", "parser", "json", "xml", "yaml", "toml"]))
df["has_input_kw"] = df["path"].apply(lambda x: has_kw(x, ["input", "request", "req", "form", "body", "param"]))
df["has_exec_kw"] = df["path"].apply(lambda x: has_kw(x, ["exec", "shell", "command", "cmd", "runtime"]))
df["has_security_kw"] = df["path"].apply(lambda x: has_kw(x, ["security", "sanitize", "sanitizer", "validate", "policy", "permission", "rbac"]))

# Агрегатные признаки по датасету
df["repo_rows_in_dataset"] = df.groupby("repo")["repo"].transform("count")
df["owner_rows_in_dataset"] = df.groupby("owner")["owner"].transform("count")
df["files_in_same_fix_commit"] = df.groupby(["owner", "repo", "fix_commit_hash"])["path"].transform("count")
df["rows_in_same_dir_in_dataset"] = df.groupby(["owner", "repo", "dir"])["path"].transform("count")
df["same_file_occurrences_in_dataset"] = df.groupby(["owner", "repo", "path"])["path"].transform("count")
df["repo_positive_ratio_in_dataset"] = df.groupby(["owner", "repo"])["label"].transform("mean")
````

````
base_feature_columns = [
    "patch_count",
    "path_depth",
    "path_len",
    "filename_len",
    "token_count_in_path",
    "raw_url_len",
    "is_test_like",
    "is_internal_path",
    "is_pkg_path",
    "is_cmd_path",
    "has_auth_kw",
    "has_crypto_kw",
    "has_network_kw",
    "has_storage_kw",
    "has_parser_kw",
    "has_input_kw",
    "has_exec_kw",
    "has_security_kw",
    "repo_rows_in_dataset",
    "owner_rows_in_dataset",
    "files_in_same_fix_commit",
    "rows_in_same_dir_in_dataset",
    "same_file_occurrences_in_dataset",
    "repo_positive_ratio_in_dataset",
]

target_column = "label"

missing = [c for c in base_feature_columns if c not in df.columns]
print("Число признаков:", len(base_feature_columns))
print("Отсутствуют:", missing)

display(df[base_feature_columns + [target_column]].head(5))
````

## 6. Анализ репозиториев и выбор нового проекта

Чтобы выполнить условие *«новый проект не использовался при обучении»*, выбираем один репозиторий целиком и исключаем его из train.

Желательно, чтобы в этом репозитории были и положительные, и отрицательные примеры — тогда можно оценить качество ранжирования.

````
repo_stats = (
    df.groupby(["owner", "repo"])["label"]
      .agg(total="size", positives="sum")
      .reset_index()
)

repo_stats["negatives"] = repo_stats["total"] - repo_stats["positives"]
repo_stats["positive_rate"] = repo_stats["positives"] / repo_stats["total"]

repo_stats = repo_stats.sort_values(["positives", "negatives", "total"], ascending=False).reset_index(drop=True)
display(repo_stats.head(20))
````

````
# Автоматический выбор holdout-проекта:
# берем репозиторий, где есть хотя бы 5 уязвимых и 20 обычных файлов.
candidates = repo_stats[(repo_stats["positives"] >= 5) & (repo_stats["negatives"] >= 20)].copy()

if len(candidates) == 0:
    raise ValueError("Не найден репозиторий, подходящий для holdout-проверки.")

holdout_owner = candidates.iloc[0]["owner"]
holdout_repo = candidates.iloc[0]["repo"]

print("Выбран новый проект для проверки:")
print(f"owner = {holdout_owner}")
print(f"repo  = {holdout_repo}")
````

При необходимости можно зафиксировать конкретный репозиторий вручную:

```python
holdout_owner = "kubernetes"
holdout_repo = "kubernetes"
```

## 7. Разделение на train и unseen-project test

````
is_holdout = (df["owner"] == holdout_owner) & (df["repo"] == holdout_repo)

df_train = df.loc[~is_holdout].copy()
df_test = df.loc[is_holdout].copy()

X_train = df_train[base_feature_columns].copy()
y_train = df_train[target_column].copy()

X_test = df_test[base_feature_columns].copy()
y_test = df_test[target_column].copy()

print("Train:", X_train.shape, y_train.shape)
print("Test :", X_test.shape, y_test.shape)
print()
print("Классы в train:")
print(y_train.value_counts())
print()
print("Классы в test:")
print(y_test.value_counts())
````

## 8. Выбор модели и обучение

Здесь используется **RandomForestClassifier**. Причины выбора:

- модель хорошо работает на табличных данных;
- не требует обязательного масштабирования признаков;
- умеет моделировать нелинейные зависимости;
- устойчива к смешанным и шумным признакам;
- дает интерпретируемость через feature importance.

Для задачи ранжирования файлов по риску это хороший базовый вариант.

````
model = RandomForestClassifier(
    n_estimators=400,
    max_depth=None,
    min_samples_leaf=2,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1,
)

model.fit(X_train, y_train)
print("Модель обучена.")
````

## 9. Предсказание вероятностей на новом проекте

````
test_proba = model.predict_proba(X_test)[:, 1]
test_pred = (test_proba >= 0.5).astype(int)

results = df_test[["owner", "repo", "path", "fix_commit_hash", "raw_url", "label"]].copy()
results["probability_vulnerability"] = test_proba
results["pred_label_05"] = test_pred
results = results.sort_values("probability_vulnerability", ascending=False).reset_index(drop=True)

display(results.head(20))
````

## 10. Оценка качества на unseen project

Смотрим несколько метрик:

- `ROC-AUC` — способность разделять классы;
- `Average Precision` — качество ранжирования для положительного класса;
- `Precision@K` и `Recall@K` — сколько реальных уязвимых файлов попадает в топ наиболее рискованных.

````
roc_auc = roc_auc_score(y_test, test_proba) if y_test.nunique() > 1 else np.nan
avg_precision = average_precision_score(y_test, test_proba) if y_test.nunique() > 1 else np.nan

print(f"ROC-AUC: {roc_auc:.4f}")
print(f"Average Precision: {avg_precision:.4f}")
print()
print("Classification report при пороге 0.5:")
print(classification_report(y_test, test_pred, digits=4))
````

````
def precision_at_k(y_true, y_score, k):
    tmp = pd.DataFrame({"y_true": y_true, "y_score": y_score}).sort_values("y_score", ascending=False)
    topk = tmp.head(k)
    return topk["y_true"].mean()

def recall_at_k(y_true, y_score, k):
    tmp = pd.DataFrame({"y_true": y_true, "y_score": y_score}).sort_values("y_score", ascending=False)
    topk = tmp.head(k)
    total_pos = tmp["y_true"].sum()
    if total_pos == 0:
        return np.nan
    return topk["y_true"].sum() / total_pos

for k in [5, 10, 20, 30, 50]:
    k_eff = min(k, len(results))
    p_at_k = precision_at_k(y_test.values, test_proba, k_eff)
    r_at_k = recall_at_k(y_test.values, test_proba, k_eff)
    print(f"K={k_eff:>3} | Precision@K={p_at_k:.4f} | Recall@K={r_at_k:.4f}")
````

## 11. Проверка: есть ли реальные уязвимые файлы среди наиболее рискованных

````
top_n = 20
top_results = results.head(top_n).copy()

print(f"Топ-{top_n} наиболее рискованных файлов:")
display(top_results[["path", "probability_vulnerability", "label"]])

top_hits = int(top_results["label"].sum())
top_precision = top_results["label"].mean()

print(f"Число реальных уязвимых файлов в top-{top_n}: {top_hits}")
print(f"Precision@{top_n}: {top_precision:.4f}")
````

````
plt.figure(figsize=(8, 4))
plt.hist(results.loc[results["label"] == 0, "probability_vulnerability"], bins=30, alpha=0.7, label="label=0")
plt.hist(results.loc[results["label"] == 1, "probability_vulnerability"], bins=30, alpha=0.7, label="label=1")
plt.title(f"Распределение вероятностей на новом проекте: {holdout_owner}/{holdout_repo}")
plt.xlabel("Предсказанная вероятность уязвимости")
plt.ylabel("Количество файлов")
plt.legend()
plt.show()
````

## 12. Важность признаков

````
feature_importance = pd.DataFrame({
    "feature": base_feature_columns,
    "importance": model.feature_importances_,
}).sort_values("importance", ascending=False).reset_index(drop=True)

display(feature_importance)

plt.figure(figsize=(8, 6))
plt.barh(feature_importance["feature"].head(15)[::-1], feature_importance["importance"].head(15)[::-1])
plt.title("Топ-15 признаков по важности")
plt.xlabel("Importance")
plt.ylabel("Feature")
plt.show()
````

## 13. Сохранение итоговой таблицы результатов

Итоговая таблица соответствует требованию задания:  
**файл, вероятность, метка при наличии**.

````
OUT_PATH = f"predictions_{holdout_owner}_{holdout_repo}.csv"
results.to_csv(OUT_PATH, index=False, encoding="utf-8-sig")
print("Сохранено:", OUT_PATH)
````

Ниже можно сформулировать вывод по результатам запуска ноутбука.

Пример структуры вывода:

1. В качестве модели была выбрана Random Forest, так как она хорошо работает на табличных признаках и не требует сложной предварительной нормализации.
2. Новый проект был выбран как полностью отложенный репозиторий, исключенный из обучения.

4. По метрикам `ROC-AUC`, `Average Precision`, `Precision@K` и `Recall@K` можно оценить, насколько модель способна ранжировать файлы по риску.
