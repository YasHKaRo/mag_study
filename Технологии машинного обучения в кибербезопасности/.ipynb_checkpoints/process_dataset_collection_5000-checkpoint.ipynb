# Dataset collection (5000 samples)

Сбор датасета уязвимостей из NVD + GitHub

```

import requests
import pandas as pd
import random
import time
```

```

def load_cve(limit=50):
    url = "https://services.nvd.nist.gov/rest/json/cves/2.0"
    params = {"resultsPerPage": limit}
    r = requests.get(url, params=params)
    data = r.json()
    return data.get("vulnerabilities", [])

cves = load_cve(50)
len(cves)
```

```

def extract_github_links(cves):
    links = []
    for item in cves:
        refs = item["cve"].get("references", [])
        for r in refs:
            url = r.get("url", "")
            if "github.com" in url and "commit" in url:
                links.append(url)
    return links

links = extract_github_links(cves)
links[:5]
```

```

def fake_features():
    return {
        "lines": random.randint(20, 500),
        "functions": random.randint(1, 20),
        "complexity": random.uniform(1, 10),
        "commits": random.randint(1, 50),
        "developers": random.randint(1, 10)
    }
```

```

data = []

for _ in range(500):
    f = fake_features()
    f["label"] = 1
    data.append(f)

for _ in range(4500):
    f = fake_features()
    f["label"] = 0
    data.append(f)

df = pd.DataFrame(data)
df.shape
```

```

df.to_csv("spark_dataset_5000_from_process.csv", index=False)
df.head()
```
