```python
import re
import time
import requests
import pandas as pd
from tqdm import tqdm

OWNER = "apache"
REPO = "spark"
PROJECT_NAME = "Apache Spark"
OUTPUT_FILE = "spark_dataset_from_process.csv"

# При желании можно добавить GitHub token, чтобы реже упираться в лимиты API
GITHUB_TOKEN = None  # например: "ghp_xxx"

session = requests.Session()
session.headers.update({"User-Agent": "dataset-builder-notebook"})
if GITHUB_TOKEN:
    session.headers.update({"Authorization": f"Bearer {GITHUB_TOKEN}"})

print(f"Репозиторий: {OWNER}/{REPO}")
print(f"Проект для NVD: {PROJECT_NAME}")
```

```python
def get_json(url, params=None, pause=0.6):
    r = session.get(url, params=params, timeout=60)
    if r.status_code == 403 and "rate limit" in r.text.lower():
        raise RuntimeError("GitHub API rate limit exceeded. Добавьте GITHUB_TOKEN.")
    r.raise_for_status()
    time.sleep(pause)
    return r.json()


def extract_english_description(descriptions):
    for item in descriptions:
        if item.get("lang") == "en":
            return item.get("value", "")
    if descriptions:
        return descriptions[0].get("value", "")
    return ""


def extract_cvss_score(cve_item):
    metrics = cve_item.get("metrics", {})
    for key in ["cvssMetricV31", "cvssMetricV30", "cvssMetricV2"]:
        if key in metrics and metrics[key]:
            data = metrics[key][0]
            cvss = data.get("cvssData", {})
            return cvss.get("baseScore")
    return None


def extract_cwe(cve_item):
    weaknesses = cve_item.get("weaknesses", [])
    for weakness in weaknesses:
        for desc in weakness.get("description", []):
            value = desc.get("value", "")
            if value.startswith("CWE-"):
                return value
    return None
```

```python
def load_cves_from_nvd(keyword=PROJECT_NAME, results_per_page=200, max_records=400):
    url = "https://services.nvd.nist.gov/rest/json/cves/2.0"
    start = 0
    rows = []

    while start < max_records:
        params = {
            "keywordSearch": keyword,
            "resultsPerPage": results_per_page,
            "startIndex": start
        }
        data = get_json(url, params=params, pause=1.0)
        vulnerabilities = data.get("vulnerabilities", [])
        if not vulnerabilities:
            break

        for item in vulnerabilities:
            cve = item.get("cve", {})
            cve_id = cve.get("id")
            desc = extract_english_description(cve.get("descriptions", []))
            score = extract_cvss_score(cve)
            cwe = extract_cwe(cve)

            refs = []
            for ref in cve.get("references", []):
                ref_url = ref.get("url", "")
                if "github.com" in ref_url.lower():
                    refs.append(ref_url)

            rows.append({
                "cve_id": cve_id,
                "published": cve.get("published"),
                "last_modified": cve.get("lastModified"),
                "description": desc,
                "cvss_score": score,
                "cwe": cwe,
                "github_refs": refs
            })

        total = data.get("totalResults", 0)
        start += results_per_page
        if start >= total:
            break

    df = pd.DataFrame(rows)
    return df


df_cves = load_cves_from_nvd()
print("Всего CVE записей из NVD:", len(df_cves))
df_cves.head(3)
```

```python
def parse_commit_url(url, owner=OWNER, repo=REPO):
    # Ищем ссылки вида github.com/<owner>/<repo>/commit/<sha>
    pattern = rf"github\.com/{re.escape(owner)}/{re.escape(repo)}/commit/([0-9a-f]{{7,40}})"
    m = re.search(pattern, url, flags=re.IGNORECASE)
    if m:
        return m.group(1)
    return None


def build_fix_commits_table(df_cves):
    rows = []
    for _, row in df_cves.iterrows():
        for ref in row["github_refs"]:
            sha = parse_commit_url(ref)
            if sha:
                rows.append({
                    "cve_id": row["cve_id"],
                    "published": row["published"],
                    "cvss_score": row["cvss_score"],
                    "cwe": row["cwe"],
                    "description": row["description"],
                    "fix_commit": sha,
                    "ref_url": ref
                })
    df = pd.DataFrame(rows).drop_duplicates(subset=["cve_id", "fix_commit"])
    return df


df_fixes = build_fix_commits_table(df_cves)
print("CVE с найденными GitHub commit-ссылками:", len(df_fixes))
df_fixes.head(10)
```

```python
def get_commit_files(owner, repo, sha):
    url = f"https://api.github.com/repos/{owner}/{repo}/commits/{sha}"
    data = get_json(url, pause=0.8)

    files = data.get("files", [])
    parents = data.get("parents", [])
    parent_sha = parents[0]["sha"] if parents else None

    rows = []
    for f in files:
        filename = f.get("filename")
        status = f.get("status")
        if not filename:
            continue
        if status == "removed":
            continue
        rows.append({
            "file_path": filename,
            "status": status,
            "additions": f.get("additions", 0),
            "deletions": f.get("deletions", 0),
            "changes": f.get("changes", 0),
            "parent_sha": parent_sha
        })
    return rows


file_rows = []
for _, row in tqdm(df_fixes.iterrows(), total=len(df_fixes)):
    try:
        changed_files = get_commit_files(OWNER, REPO, row["fix_commit"])
        for item in changed_files:
            file_rows.append({
                "cve_id": row["cve_id"],
                "fix_commit": row["fix_commit"],
                "published": row["published"],
                "cvss_score": row["cvss_score"],
                "cwe": row["cwe"],
                "description": row["description"],
                **item
            })
    except Exception as e:
        print("Ошибка на commit", row["fix_commit"], e)

df_changed = pd.DataFrame(file_rows)
print("Файлов из исправляющих коммитов:", len(df_changed))
df_changed.head(10)
```

```python
# Оставим только исходный код. Это упрощает датасет и делает его ближе к заданию.
CODE_EXTENSIONS = (
    ".java", ".scala", ".py", ".c", ".cpp", ".cc", ".h", ".hpp",
    ".js", ".ts", ".go", ".rb"
)

df_changed = df_changed[df_changed["file_path"].str.endswith(CODE_EXTENSIONS, na=False)].copy()
df_changed = df_changed.drop_duplicates(subset=["fix_commit", "file_path"])
print("Файлов исходного кода после фильтрации:", len(df_changed))
df_changed.head(10)
```

```python
def download_file_at_commit(owner, repo, path, ref):
    url = f"https://raw.githubusercontent.com/{owner}/{repo}/{ref}/{path}"
    r = session.get(url, timeout=60)
    if r.status_code == 404:
        return None
    r.raise_for_status()
    time.sleep(0.5)
    return r.text


def count_functions(code, file_path):
    ext = "." + file_path.split(".")[-1].lower() if "." in file_path else ""
    if ext in [".py"]:
        pattern = r"^\s*def\s+[A-Za-z_][A-Za-z0-9_]*\s*\("
    elif ext in [".java", ".scala", ".js", ".ts", ".go", ".c", ".cpp", ".cc", ".h", ".hpp"]:
        pattern = r"\b[A-Za-z_][A-Za-z0-9_<>,\[\]\s\*]*\s+[A-Za-z_][A-Za-z0-9_]*\s*\([^;]*\)\s*\{"
    else:
        pattern = r"\bdef\b|\bfunction\b"
    return len(re.findall(pattern, code, flags=re.MULTILINE))


def simple_code_metrics(code, file_path):
    if not code or not code.strip():
        return None

    lines = code.splitlines()
    non_empty = [line for line in lines if line.strip()]
    comment_lines = [line for line in lines if line.strip().startswith(("//", "#", "/*", "*"))]

    loc = len(non_empty)
    comment_count = len(comment_lines)
    function_count = count_functions(code, file_path)

    complexity_words = re.findall(r"\b(if|for|while|case|catch|except|elif|switch)\b", code)
    import_words = re.findall(r"\b(import|from|include|package)\b", code)
    class_words = re.findall(r"\b(class|interface|object)\b", code)

    return {
        "loc": loc,
        "comment_lines": comment_count,
        "comment_ratio": round(comment_count / loc, 4) if loc else 0,
        "function_count": function_count,
        "control_flow_count": len(complexity_words),
        "import_count": len(import_words),
        "class_count": len(class_words)
    }
```

```python
def get_file_commit_stats(owner, repo, path, until_sha=None):
    # Для простоты берём только последние коммиты по файлу.
    # Этого достаточно для учебной практической.
    url = f"https://api.github.com/repos/{owner}/{repo}/commits"
    params = {"path": path, "per_page": 30}
    data = get_json(url, params=params, pause=0.8)

    authors = set()
    dates = []

    for item in data:
        commit = item.get("commit", {})
        author = commit.get("author", {})
        if author.get("name"):
            authors.add(author["name"])
        if author.get("date"):
            dates.append(author["date"][:10])

    commits_count = len(data)
    unique_authors = len(authors)

    if dates:
        dates = sorted(dates)
        first_date = dates[0]
        last_date = dates[-1]
    else:
        first_date = None
        last_date = None

    return {
        "recent_commits_for_file": commits_count,
        "unique_authors": unique_authors,
        "first_seen_date": first_date,
        "last_seen_date": last_date
    }
```

```python
dataset_rows = []

sample_limit = 120  # чтобы ноутбук не работал слишком долго
work_df = df_changed.head(sample_limit).copy()

for _, row in tqdm(work_df.iterrows(), total=len(work_df)):
    try:
        code_before_fix = download_file_at_commit(OWNER, REPO, row["file_path"], row["parent_sha"])
        metrics = simple_code_metrics(code_before_fix, row["file_path"])
        if not metrics:
            continue

        hist = get_file_commit_stats(OWNER, REPO, row["file_path"])

        dataset_rows.append({
            "cve_id": row["cve_id"],
            "file_path": row["file_path"],
            "fix_commit": row["fix_commit"],
            "parent_sha": row["parent_sha"],
            "published": row["published"],
            "cvss_score": row["cvss_score"],
            "cwe": row["cwe"],
            "is_vulnerable": 1,
            "additions": row["additions"],
            "deletions": row["deletions"],
            "changes": row["changes"],
            **metrics,
            **hist
        })
    except Exception as e:
        print("Ошибка на файле", row["file_path"], e)

df_vuln = pd.DataFrame(dataset_rows)
print("Уязвимых объектов:", len(df_vuln))
df_vuln.head(10)
```

```python
def build_safe_examples(df_vuln, target_ratio=3):
    # Формируем простые "безопасные" объекты:
    # берём файлы из тех же исправляющих коммитов, но без привязки к CVE,
    # либо дублируем часть истории из других файлов как baseline.
    # Чтобы не усложнять, используем существующие записи и меняем метку
    # только для файлов, которые не фигурируют как уязвимые.
    vuln_files = set(df_vuln["file_path"].unique())
    safe_candidates = df_changed[~df_changed["file_path"].isin(vuln_files)].copy()

    rows = []
    need = min(len(safe_candidates), len(df_vuln) * target_ratio)

    for _, row in tqdm(safe_candidates.head(need).iterrows(), total=min(len(safe_candidates), need)):
        try:
            code = download_file_at_commit(OWNER, REPO, row["file_path"], row["parent_sha"])
            metrics = simple_code_metrics(code, row["file_path"])
            if not metrics:
                continue

            hist = get_file_commit_stats(OWNER, REPO, row["file_path"])

            rows.append({
                "cve_id": None,
                "file_path": row["file_path"],
                "fix_commit": row["fix_commit"],
                "parent_sha": row["parent_sha"],
                "published": row["published"],
                "cvss_score": None,
                "cwe": None,
                "is_vulnerable": 0,
                "additions": row["additions"],
                "deletions": row["deletions"],
                "changes": row["changes"],
                **metrics,
                **hist
            })
        except Exception as e:
            print("Ошибка на safe-файле", row["file_path"], e)

    return pd.DataFrame(rows)


df_safe = build_safe_examples(df_vuln, target_ratio=2)
print("Безопасных объектов:", len(df_safe))
df_safe.head(10)
```

```python
df_final = pd.concat([df_vuln, df_safe], ignore_index=True)
df_final = df_final.drop_duplicates(subset=["file_path", "parent_sha", "is_vulnerable"])
df_final = df_final.reset_index(drop=True)

print("Итоговый размер датасета:", len(df_final))
print(df_final["is_vulnerable"].value_counts(dropna=False))
df_final.head(10)
```

```python
# Небольшая предобработка
df_final["published"] = pd.to_datetime(df_final["published"], errors="coerce")
df_final["first_seen_date"] = pd.to_datetime(df_final["first_seen_date"], errors="coerce")
df_final["last_seen_date"] = pd.to_datetime(df_final["last_seen_date"], errors="coerce")

df_final["file_age_days"] = (df_final["last_seen_date"] - df_final["first_seen_date"]).dt.days
df_final["file_age_days"] = df_final["file_age_days"].fillna(0)

df_final = df_final.fillna({
    "cvss_score": 0,
    "cwe": "UNKNOWN",
    "comment_ratio": 0,
    "file_age_days": 0
})

df_final.to_csv(OUTPUT_FILE, index=False)
print("Файл сохранён:", OUTPUT_FILE)
print("Столбцы датасета:")
print(list(df_final.columns))
```
