```python
# AlexNet
def net ():
    return tf.keras.models.Sequential([tf.keras.layers.Conv2D(filters=96, kernel_size=11, strides=4,activation='relu'),
# первый сверточный слой
    tf.keras.layers.MaxPool2D(pool_size=3, strides=2),
# субдискретизация
# мешьшее ядро 5 х 5, но большее количество
    tf.keras.layers.Conv2D(filters=256, kernel_size=5, padding='same',activation='relu'),
# второй сверточный слой
    tf.keras.layers.MaxPool2D(pool_size=3, strides=2),
# субдискретизация
# сверточные слои с малым ядром 3 х 3, с набивкой
    tf.keras.layers.Conv2D(filters=384, kernel_size=3, padding='same',activation='relu'),
    tf.keras.layers.Conv2D(filters=384, kernel_size=3, padding='same',activation='relu'),
# последний сверточный слой с 256 фильтрами, малым
    tf.keras.layers.Conv2D(filters=256, kernel_size=3, padding='same',activation='relu'),
    tf.keras.layers.MaxPool2D(pool_size=3, strides=2),
# субдискретизация
    tf.keras.layers.Flatten(),
# слой, который вытягивает входной массив в вектор
# полносвязный слой
    tf.keras.layers.Dense(4096, activation='relu'),
    tf.keras.layers.Dropout(0.5),
# для лучшего обобщения
# полносвязный слой
    tf.keras.layers.Dense(4096, activation='relu'),
    tf.keras.layers.Dropout(0.5),
# для лучшего обобщения
# последний полносвязный слой с числом нейронов по
    tf.keras.layers.Dense(10)])
    
```
